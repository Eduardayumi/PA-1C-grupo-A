```python
# 06- Considere as notas = {7.5, 8.0, 6.5, 9.0, 5.5} Mostre a maior nota, depois a menor, depois a médias das notas

notas = [7.5, 8.0, 6.5, 9.0, 5.5]

maior_nota = max(notas)
menor_nota = min(notas)
media_notas = sum(notas) / len(notas)

print(f"Maior nota: {maior_nota}")
print(f"Menor nota: {menor_nota}")
print(f"Média da nota: {media_notas}")



```

```python
# 09- Crie uma lista chamada números. Peça ao usuario para informar 10 números, depois crie duas listas vazias chamada
#pares e impares. Percorra a lista original e coloque cada número na lista correspondente
numeros = []
pares = []
impares = []

for i in range(10):
  cu = int(input("Digite um número: "))
  numeros.append(cu)

for i in numeros:
    if i % 2 == 0:
        pares.append(i)
    else:
        impares.append(i)

print("Lista original:", numeros)
print("Lista de pares:", pares)
print("Lista de ímpares:", impares)
```

```python
# 10- Faça um pequeno sistema para cadastrar as notas de 05 alunos; Programa deverá armazenar as notas de uma lista
# e depois mostrar todas as notas, a maior nota, média da nota, quantidade de alunos com nota maiores ou igual a 6, quantidade de alunos com notas menores que 6

notas = []
acima_ou_igual_6 = 0
abaixo_6 = 0

for i in range(5):
    nota = float(input(f"Digite a nota do {i+1}º aluno: "))
    notas.append(nota)

media = sum(notas) / len(notas)
maior_nota = max(notas)

for i in notas:
    if i >= 6.0:
        acima_ou_igual_6 += 1
    else:
        abaixo_6 += 1

print("Resultados")
print(f"Notas cadastradas: {notas}")
print(f"Maior nota: {maior_nota}")
print(f"Média da turma: {media:.2f}")
print(f"Alunos com nota maior ou igual a 6: {acima_ou_igual_6}")
print(f"Alunos com nota menor que 6: {abaixo_6}")
```
